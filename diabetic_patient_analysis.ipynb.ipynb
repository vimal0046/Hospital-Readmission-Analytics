```python
ct = pd.crosstab(df["age"], df["readmitted"])

ct.plot(kind="bar", stacked=True, figsize=(10,6))

plt.title("Hospital Readmission by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Patients")
plt.show()
```

Observation

Patients in older age groups (60–80) show higher hospital readmission counts.

Younger patients have relatively fewer readmissions.

This suggests age may be an important factor influencing hospital readmission.

```python
df = df[df["gender"] != "Unknown/Invalid"] # remove unknown values
```

```python
pd.crosstab(df["gender"],df["readmitted"])
```

```python
ct = pd.crosstab(df["gender"], df["readmitted"])

ct.plot(kind="bar", stacked=True, figsize=(8,5))

plt.title("Hospital Readmission by gender")
plt.xlabel("gender")
plt.ylabel("Number of Patients")
plt.show()
```

Observation

Male and female patients show similar readmission patterns.

```python
pd.crosstab(df["race"],df["readmitted"])
```

```python
ct= pd.crosstab(df["race"],df["readmitted"])
ct = ct.sort_values(by="NO", ascending=False)
ct.plot(kind = "bar",stacked = True,figsize = (10,6))
plt.title("Hospital Readmission by race")
plt.xlabel("race")
plt.ylabel("Number of Patients")
plt.show()
```

Observation:

Caucasian patients represent the largest group in the dataset.

Other racial groups appear in smaller proportions.

Readmission patterns vary slightly across racial groups.
